### Text Preprocessing

```python
import regex as re
```

```python
# remove html tags
def remove_tags(text):
    pattern = "<.*?>"    
    return re.sub(pattern,'',text)
```

```python
new_df['review'] = new_df['review'].apply(remove_tags)
```

```python
new_df.iloc[2].review
```

```python
def convert_lower(text):
    return text.lower()
```

```python
new_df['review']=new_df['review'].apply(convert_lower)
```

```python
new_df.iloc[3].review
```

```python
#to remove special characters
def remove_special(text):
    pattern = re.compile("[$&+,:;=?@#|'<>.-^*()%!]")
    return re.sub(pattern,'',text)
```

```python
new_df['review']=new_df['review'].apply(remove_special)
```

```python
new_df.iloc[3].review
```

```python
# stop word removal
import nltk
```

```python
from nltk.corpus import stopwords
```

```python
def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))
    processed = [word for word in text.split() if word not in stop_words]
    return processed
```

```python
remove_stopwords(new_df.iloc[3].review)
```

```python
new_df['review']=new_df['review'].apply(remove_stopwords)
```

```python
new_df
```

```python
new_df.iloc[0].review
```

```python
# perform stemming
from nltk.stem import PorterStemmer
ps = PorterStemmer()
def convert_stem(text):
    stemmed = [ps.stem(i) for i in text]
    return stemmed
```

```python
temp = convert_stem(new_df.iloc[3].review)
```

```python
len(new_df.iloc[3].review), len(temp)
```

```python
stemmed_df = pd.DataFrame()
```

```python
stemmed_df['review']=new_df["review"].apply(convert_stem)
```

```python
stemmed_df['sentiment'] = new_df['sentiment']
```

```python
stemmed_df.head()
```

```python
stemmed_df['sentiment'] = stemmed_df['sentiment'].replace({'positive':1, 'negative':0},inplace=True)
```

```python
stemmed_df['sentiment']
```

```python
# join back
def join_text(stem_list):
    return ' '.join(stem_list)
```

```python
stemmed_df['review']=stemmed_df['review'].apply(join_text)
```

```python
stemmed_df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```
