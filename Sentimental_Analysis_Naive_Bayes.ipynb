### Text Preprocessing

```python
import regex as re
```

```python
# remove html tags
def remove_tags(text):
    pattern = "<.*?>"    
    return re.sub(pattern,'',text)
```

```python
new_df['review'] = new_df['review'].apply(remove_tags)
```

```python
new_df.iloc[2].review
```

```python
def convert_lower(text):
    return text.lower()
```

```python
new_df['review']=new_df['review'].apply(convert_lower)
```

```python
new_df.iloc[3].review
```

```python
#to remove special characters
def remove_special(text):
    pattern = re.compile("[$&+,:;=?@#|'<>.-^*()%!]")
    return re.sub(pattern,'',text)
```

```python
new_df['review']=new_df['review'].apply(remove_special)
```

```python
new_df.iloc[3].review
```

```python
# stop word removal
import nltk
```

```python
from nltk.corpus import stopwords
```

```python
def remove_stopwords(text):
    clean = [i for i in text.split(' ') if i not in stopwords.words('english') and i!='']
    return clean
```

```python
remove_stopwords(new_df.iloc[3].review)
```

```python
new_df['review']=new_df['review'].apply(remove_stopwords)
```

```python
new_df
```

```python
new_df.iloc[0].review
```

```python
# perform stemming
from nltk.stem import PorterStemmer
ps = PorterStemmer()
def convert_stem(text):
    stemmed = [ps.stem(i) for i in text]
    return stemmed
```

```python
temp = convert_stem(new_df.iloc[3].review)
```

```python
len(new_df.iloc[3].review), len(temp)
```

```python
new_df['review']=new_df["review"].apply(convert_stem)
```

```python
# join back
def join_text(stem_list):
    return ' '.join(stem_list)
```

```python
new_df['review']=new_df['review'].apply(join_text)
```

```python
new_df
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

```python
X=cv.fit_transform(new_df['review']).toarray()
```

## ML part

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# Algorithm: Naive Bayes
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
```

```python
clf1 = GaussianNB()
clf2 = MultinomialNB()
clf3 = BernoulliNB()
```

```python
clf1.fit(X_train,y_train)
clf2.fit(X_train,y_train)
clf3.fit(X_train,y_train)
```

```python
y_pred_1 = clf1.predict(X_test)
y_pred_2 = clf2.predict(X_test)
y_pred_3 = clf3.predict(X_test)
```

```python
y_pred_1.shape
```

```python
# Accuracy
from sklearn.metrics import accuracy_score
```

```python
print("Gaussian: ", accuracy_score(y_test, y_pred_1))
print("Multinomial: ", accuracy_score(y_test, y_pred_2))
print("Bernaulli: ", accuracy_score(y_test, y_pred_3))
```
